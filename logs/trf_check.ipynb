```python
import matplotlib.pyplot as plt
import numpy as np

with open("lg_3_4_missed.txt", 'r') as in_file:
    text_4 = in_file.read().split("\n")

with open("lg_3_8_missed.txt", 'r') as in_file:
    text_8 = in_file.read().split("\n")

text_4 = [ t.lower() for t in text_4 if t != '' ]
text_8 = [ t.lower() for t in text_8 if t != '' ]

p_set_4 = list(set(text_4))
counts_4 = []
for p in p_set_4:
    counts_4.append((p, text_4.count(p)))
counts_4.sort(key=lambda x: x[0])
print(counts_4)

p_set_8 = list(set(text_8))
counts_8 = []
for p in p_set_8:
    counts_8.append((p, text_8.count(p)))
counts_8.append(('itself', 0))
counts_8.sort(key=lambda x: x[0])
print(counts_8)

# Make fast bar chart

pronouns = [c[0] for c in counts_4]
c_4 = [c[1] for c in counts_4]
c_8 = [c[1] for c in counts_8]

fig, ax1 = plt.subplots(figsize=(8, 8))

w, x = 0.4, np.arange(len(pronouns))

ax1.bar(x - w/2, c_4, width=w, label="lg_3_4")
ax1.bar(x + w/2, c_8, width=w, label="lg_3_8")
ax1.set_xlabel("Pronoun type")
ax1.set_xticks(x)
ax1.set_xticklabels(pronouns)
ax1.set_ylabel("Total Errors")
ax1.set_title("Comparing spacy models test error")
ax1.legend()

plt.xticks(rotation=45)
plt.show()
```
